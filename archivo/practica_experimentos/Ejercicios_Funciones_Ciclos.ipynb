## 🥊 Reto 1: Tu Primera Fábrica (Funciones)
Piensa en una función como una pequeña maquinita: le metes un ingrediente crudo (el parámetro), la máquina hace algo con él por dentro, y te devuelve un producto terminado (`return`).

**Misión:** Crea una función llamada `saludar_estudiante` que reciba un `nombre` como parámetro, y que **devuelva** (con `return`) un mensaje de texto que diga `"¡Hola, [nombre]!"`.

- Para declarar una función usas la palabra secreta `def`.
- Recuerda los dos puntos `:` al final de la primera línea y la **sangría** (el espacio a la izquierda) en las líneas de abajo.
- Para mezclar texto con variables de forma elegante, usa las `f-strings`: `f"¡Hola, {nombre}!"`

```python
# Escribe tu código aquí para el Reto 1:
def saludar_estudiante(nombre):
    mensaje = f"Hola {nombre}"
    pass
    return mensaje

# Ejecutamos la máquina para probarla:
print(saludar_estudiante("Carlos"))
```

## 🥊 Reto 2: El Profesor Estricto (Condicionales)
**Misión:** Crea una función llamada `calificar_nota` que reciba una `nota` (que será un número). Si la nota es mayor o igual a 80, la máquina debe **devolver** la palabra `"Aprobado"`. Si es menor, debe devolver `"Reprobado"`.

💡 **Pistas:** 
- Adentro de la función, usa los condicionales lógicos `if` y `else`.
- No olvides los dos puntos `:` al final del `if`.
- Ejemplo de estructura lógica: `if nota >= 80:`

```python
# Escribe tu código aquí para el Reto 2:
def calificar_nota(nota):
    if nota >=80:
        estado = "Aprobado"
    else:
        estado = 'Reprobado'
    
    return estado



# Pruebas (Esto se encarga de probar si tu código funcionó):
print("Nota 85:", calificar_nota(85)) # Debería imprimir Aprobado
print("Nota 60:", calificar_nota(60)) # Debería imprimir Reprobado
```

## 🥊 Reto 3: El Pase de Lista (Ciclo For)
Un ciclo `for` nos permite decirle a Python: *"Toma esta lista enorme y pásame cada elemento, uno por uno, para que yo le haga algo"*.

**Misión:** Tienes una lista de estudiantes. Escribe un ciclo `for` que pase por cada uno de ellos y lo imprima en la pantalla de forma automática.

💡 **Pistas:** 
- La estructura mágica es: `for estudiante in lista_estudiantes:`
- Recuerda que todo lo que va ADENTRO del ciclo debe llevar un pequeño espacio a la izquierda (sangría).

```python
lista_estudiantes = ["Ana", "Luis", "María", "Juan"]

# Escribe tu código aquí para el Reto 3:
for estudiante in lista_estudiantes:
    print("El estudiante es: ",estudiante)
```

## 🥊 Reto 4: El Gran Final (Mezclando el poder de ambos)
**Misión:** Tienes una lista de notas peladas. Escribe un ciclo `for` que pase por cada nota de la lista y **llame a tu función del Reto 2** (`calificar_nota`) para imprimir en pantalla si esa nota fue Aprobada o Reprobada.

💡 **Pistas:** 
- Primero, creas tu ciclo para recorrer la lista: `for nota in lista_notas:`
- Adentro del ciclo, llamas a tu función y la imprimes: `print(calificar_nota(nota))`.
- ¡Si esto te sale bien, oficialmente sabrás automatizar procesos!

```python
lista_notas = [45, 90, 80, 75, 100]

for nota in lista_notas:
    print(calificar_nota(nota))
```
