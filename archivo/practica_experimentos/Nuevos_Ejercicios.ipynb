### Ejercicio 2: Filtrar los Pares
Escribe una función llamada `filtrar_pares(numeros)` que reciba una lista de números enteros. La función debe usar un ciclo para crear una **nueva lista** que contenga únicamente los números pares de la lista original, y luego retornar esa nueva lista.
* **Ejemplo:** `filtrar_pares([1, 2, 3, 4, 5, 6])` debe retornar `[2, 4, 6]`.

````python
# Tu código aquí

def filtrar_pares(numeros):
    lista_par = []
    for numero in numeros:
        if numero % 2 == 0:
            lista_par.append(numero)
    return lista_par
            

# Prueba tu función:
print(filtrar_pares([1, 2, 3, 4, 5, 6]))
````

### Ejercicio 3: Factorial de un número
Crea una función llamada `calcular_factorial(n)` que reciba un número entero positivo. Usa un ciclo (`for` o `while`) para calcular su factorial. (Recuerda que el factorial de 5, por ejemplo, es `5 * 4 * 3 * 2 * 1 = 120`).
* Si el número es 0, la función debe retornar 1.

````python


def calcular_factorial(n):
    resultado = 1
    if n == 0:
        resultado = 1
    for i in  range(1, n + 1):
     resultado = resultado * i
    return resultado

print(calcular_factorial(5))
````
